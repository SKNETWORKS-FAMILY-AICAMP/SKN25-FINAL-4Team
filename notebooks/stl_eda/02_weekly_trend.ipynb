```python
import sys
from pathlib import Path

ROOT = Path('/home/aceya/EMS')
sys.path.insert(0, str(ROOT / 'src'))

import pandas as pd
import numpy as np
import plotly.graph_objects as go
from plotly.subplots import make_subplots

from ems.db import load_env, connect

load_env()
```

```python
# 대상 계량기 정의
COOLING_ELECTRIC = ['H1.Z16', 'H1.Z11', 'H1.Z12', 'H1.Z24', 'H1.Z25']
COOLING_THERMAL  = ['V.K21', 'H1.K11', 'H1.K12', 'H1.K14', 'H1.K15', 'H1.K16', 'H2.K21']
HEATING_ELECTRIC = ['H1.Z20', 'H1.ZE20']
HEATING_THERMAL  = ['H1.W11', 'H1.W12']

ALL_METERS = COOLING_ELECTRIC + COOLING_THERMAL + HEATING_ELECTRIC + HEATING_THERMAL

# 조회 기간: 2023년 12월
START = '2023-12-01'
END   = '2024-01-01'

print(f'전체 계량기 수: {len(ALL_METERS)}')
print(f'조회 기간: {START} ~ {END}')
```

```python
def fetch_meter_data(meter_urn: str) -> pd.DataFrame:
    sql = """
        SELECT
            time_bucket('1 hour', ts) AS ts,
            measurement,
            AVG(value) AS value
        FROM ems.cr_measurement_1h
        WHERE meter_urn = %s
          AND ts >= %s
          AND ts <  %s
        GROUP BY 1, measurement
        ORDER BY 1, measurement
    """
    with connect() as conn:
        df = pd.read_sql(sql, conn, params=(meter_urn, START, END))

    if df.empty:
        return pd.DataFrame()

    df = df.pivot(index='ts', columns='measurement', values='value')
    df.index = pd.to_datetime(df.index, utc=True).tz_convert('Europe/Berlin')
    df.index.name = 'timestamp'
    return df
```

```python
import os

save_dir_csv = ROOT / 'outputs/tables/stl_eda/weekly'
os.makedirs(save_dir_csv, exist_ok=True)

for meter in ALL_METERS:
    df = fetch_meter_data(meter)
    if df.empty:
        print(f'{meter} 데이터 없음')
        continue

    df['weekday'] = df.index.weekday          # 0=월 ~ 6=일
    df['week']    = df.index.isocalendar().week.astype(int)
    df['hour']    = df.index.hour

    # 요일별 평균 (주차별 분리)
    weekly_mean = df.groupby(['week', 'weekday']).mean(numeric_only=True)
    weekly_mean.to_csv(save_dir_csv / f'{meter}_weekly_mean.csv')
    print(f'{meter} 저장 완료')
```

```python
WEEKDAY_LABELS = ['월', '화', '수', '목', '금', '토', '일']
COLORS = ['#1f77b4', '#ff7f0e', '#2ca02c', '#d62728', '#9467bd']

def plot_weekly_trend(meter_urn: str, df: pd.DataFrame, title_suffix: str = '') -> go.Figure:
    df = df.reset_index()
    measurements = [c for c in df.columns if c not in ['week', 'weekday']]
    n_meas = len(measurements)
    ncols = min(4, n_meas)
    nrows = (n_meas + ncols - 1) // ncols

    fig = make_subplots(
        rows=nrows, cols=ncols,
        subplot_titles=measurements,
        shared_xaxes=False,
    )

    weeks = sorted(df['week'].unique())

    for idx, meas in enumerate(measurements):
        row = idx // ncols + 1
        col = idx % ncols + 1

        for w_idx, week in enumerate(weeks):
            week_data = df[df['week'] == week].sort_values('weekday')
            fig.add_trace(
                go.Scatter(
                    x=week_data['weekday'],
                    y=week_data[meas],
                    name=f'{week}주차',
                    legendgroup=str(week),
                    showlegend=(idx == 0),
                    line=dict(color=COLORS[w_idx % len(COLORS)]),
                    mode='lines+markers',
                    marker=dict(size=5),
                ),
                row=row, col=col
            )

        fig.update_xaxes(
            tickvals=list(range(7)),
            ticktext=WEEKDAY_LABELS,
            row=row, col=col
        )

    fig.update_layout(
        title=f'{meter_urn} 주간 트렌드 (요일별 평균, 2023년 12월) {title_suffix}',
        height=max(300, nrows * 250),
        width=1400,
        legend=dict(title='주차'),
        template='plotly_white',
    )
    return fig
```

```python
save_dir_png = ROOT / 'outputs/figures/stl_eda/weekly'
os.makedirs(save_dir_png, exist_ok=True)

for meter in ALL_METERS:
    csv_path = save_dir_csv / f'{meter}_weekly_mean.csv'
    if not csv_path.exists():
        print(f'{meter} CSV 없음')
        continue

    df = pd.read_csv(csv_path)
    fig = plot_weekly_trend(meter, df, '')
    fig.write_image(
        str(save_dir_png / f'{meter}_weekly_trend.png'),
        width=1400, height=800
    )
    print(f'{meter} 저장 완료')
```
